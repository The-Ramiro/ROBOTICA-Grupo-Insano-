```python
import numpy as np
# Ejercicio_3: Implementación y Validación en Python (NumPy)

#Defina las funciones rot_x(deg), rot_y(deg) y rot_z(deg).
def rot_x(deg):
    rad = np.radians(deg)
    return np.array([
        [1, 0, 0],
        [0, np.cos(rad), -np.sin(rad)],
        [0, np.sin(rad),  np.cos(rad)]
    ])
def rot_y(deg):
    rad = np.radians(deg)
    return np.array([
        [ np.cos(rad), 0, np.sin(rad)],
        [ 0,           1, 0          ],
        [-np.sin(rad), 0, np.cos(rad)]
    ])

def rot_z(deg):
    rad = np.radians(deg)
    return np.array([
        [np.cos(rad), -np.sin(rad), 0],
        [np.sin(rad),  np.cos(rad), 0],
        [0,            0,           1]
    ])

# Cálculo numérico de la matriz ARB
A_R_B = rot_y(-45) @ rot_x(45) @ rot_z(90)
print("2. Matriz A_R_B Resultante:")
print ()
print(np.round(A_R_B, 4))
print()

# Comprobación mediante la norma de Frobenius
I = np.eye(3)
norma_frobenius = np.linalg.norm(A_R_B.T @ A_R_B - I, ord='fro')

print("3. Verificación de Ortogonalidad:")
print()
print(f"||R^T * R - I||_F = {norma_frobenius:.2e}")
print(f"¿Cumple ||R^T * R - I||_F < 10^-15?: {norma_frobenius < 1e-15}")
print()

# Transformación de Vector e Isometría
p_B = np.array([2.0, -1.0, 4.0])
p_A = A_R_B @ p_B

norma_p_B = np.linalg.norm(p_B)
norma_p_A = np.linalg.norm(p_A)
diferencia_normas = abs(norma_p_A - norma_p_B)

print("4. Transformación de Vector e Isometría ")
print()
print(f"p_B = {p_B}")
print(f"p_A = {np.round(p_A, 4)}")
print(f"||p_B|| = {norma_p_B:.6f}")
print(f"||p_A|| = {norma_p_A:.6f}")
print(f"||p_A|| - ||p_B|| = {diferencia_normas:.2e}")
print(f"¿Se preserva la norma?: {np.isclose(diferencia_normas, 0.0)}")
```
